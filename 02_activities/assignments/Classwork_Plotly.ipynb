```python
import pandas as pd
import matplotlib.pyplot as plt
from wordcloud import WordCloud

# Load dataset
df = pd.read_csv(
    "https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv",
    on_bad_lines='skip'
)

# Join all quotes into one big string
text = " ".join(each for each in df['quote'].astype(str))

# Generate word cloud
wordcloud = WordCloud(
    background_color="white",
    colormap='inferno'
).generate(text)

# Display it
fig, ax = plt.subplots(figsize=(7, 3))
ax.imshow(wordcloud, interpolation='bilinear')
ax.axis("off")

plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib_venn import venn2_unweighted

A = {"apple", "banana", "watermelon"}
B = {"pumpkin", "blueberry", "apple", "key lime"}

diagram = venn2_unweighted(
    [A, B],
    set_labels=('Fruits', 'Pies'),
    set_colors=("blue", "red"),
    alpha=0.5
)

# ✅ keep .set_text on the same line and use "\n" for newlines
diagram.get_label_by_id("10").set_text("\n".join(A - B))
diagram.get_label_by_id("11").set_text("\n".join(A & B))
diagram.get_label_by_id("01").set_text("\n".join(B - A))

plt.title("Fruit and Pie Ingredient Overlap")
plt.show()
```
