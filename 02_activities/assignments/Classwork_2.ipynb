```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])   # set x-axis from 0–10 and y-axis from 0–10

# Text in data coordinates (x=1, y=5)
ax.text(1, 5, ". Data:(1, 5)", transform=ax.transData)

# Text in axes coordinates (0–1 range inside the Axes box)
ax.text(0.5, 0.1, ". Axes:(0.5, 0.1)", transform=ax.transAxes)

# Text in figure coordinates (0–1 range for the entire Figure window)
ax.text(0.2, 0.2, ". Figure:(0.2, 0.2)", transform=fig.transFigure)

plt.show()

ax.text(1, 5, ". Data: (1, 5)",           transform=ax.transData    )
ax.text(0.5, 0.1, ". Axes: (0.5, 0.1)",   transform=ax.transAxes    )
ax.text(0.2, 0.2, ". Figure: (0.2, 0.2)", transform=fig.transFigure )
```

```python
import matplotlib.pyplot as plt

# Define your data first
x = [5, 10, 15, 20, 25]
y1 = [80, 95, 85, 90, 100]
y2 = [75, 92, 88, 85, 95]

# Create the figure and axes
fig, ax = plt.subplots(figsize=(5, 3))

# Scatter plots
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")

# Add legend
ax.legend(loc='lower right')

# Add annotation with an arrow
ax.annotate(
    'This is important!',
    xy=(10, 95),            # point to mark
    xytext=(20, 94),        # text position
    arrowprops=dict(facecolor='black', arrowstyle='->')
)

plt.show()

ax.annotate('This is important!',
            xy=(10, 95),
            xytext=(20, 94),
            arrowprops=dict(arrowstyle="wedge", color="hotpink"))
```

```python
import matplotlib.pyplot as plt

# Example data
x = [5, 10, 15, 20, 25]
y1 = [80, 95, 85, 90, 100]
y2 = [75, 92, 88, 85, 95]

# ✅ No spaces in the dictionary keys
font1 = {'family': 'serif', 'color': 'indigo'}

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')

# Use fontdict properly
plt.xlabel('Shiny New X Axis!', fontsize=18, fontdict=font1)
plt.show()
```
