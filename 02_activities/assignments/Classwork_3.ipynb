```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

np.random.seed(613)  # For reproducibility
```

```python
tips = sns.load_dataset("tips")
print(tips)
```

```python
sns.lineplot(data=tips,       # choose our dataset
           x='total_bill',  # define our x variable
           y='tip')         # define our y variable
```

```python
sns.set_style('whitegrid')

tipgraph = sns.lineplot(data=tips,
                        x='total_bill',
                        y='tip')

tipgraph.set(title='Tips vs. Total Bill', 
            xlabel='Total Bill ($)', 
            ylabel='Tip Amount ($)')


tipgraph = sns.lineplot(data=tips,
                      x='total_bill',
                      y='tip',
                      color = 'hotpink',
                      linestyle = '--',
                      linewidth = 3,
                      marker = 'o',
                      markerfacecolor = 'indigo')
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Example dataset
tips = sns.load_dataset('tips')

# Scatterplot
tipgraph = sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',           # ✅ straight quotes
    style='time',
    hue='day',
    palette=['purple', 'hotpink', 'deepskyblue', 'yellowgreen']
)

# Add title and labels
tipgraph.set(
    title='Tips vs. Total Bill',
    xlabel='Total Bill ($)',
    ylabel='Tip Amount ($)'
)

plt.show()
```

```python
sns.pairplot(
      data = tips,
      hue = 'day')
```

```python
daysplot = sns.relplot(
            data=tips,
            x="total_bill",
            y="tip",
            hue="sex",
            col="day",
            kind="scatter",
            col_wrap=2)
```
